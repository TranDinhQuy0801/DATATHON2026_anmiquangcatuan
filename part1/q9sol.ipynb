# Question 9
file: order_items, returns, products
cand: size
target: count(return)/count(order_items)
calc: max

```python
import pandas as pd

df_rtn = pd.read_csv('../dataset/returns.csv')
df_ord = pd.read_csv('../dataset/order_items.csv' ,low_memory=False)
df_prd = pd.read_csv('../dataset/products.csv')

#join with product_id -> orders and sizes -> group and count
df_order_with_size = pd.merge(df_ord, df_prd[['product_id', 'size']], on='product_id', how='inner')
total_order_with_size = df_order_with_size.groupby('size').size().reset_index(name = 'row_counts_ord')
#join with product_id -> orders and sizes -> group and count
df_returns_with_size = pd.merge(df_rtn, df_prd[['product_id', 'size']], on='product_id', how='inner')
total_return_with_size = df_returns_with_size.groupby('size').size().reset_index(name = 'row_counts_rtn')
#join and calc
df_final = pd.merge(total_order_with_size, total_return_with_size, on='size', how='left').fillna(0)
df_final['return_rate'] = df_final['row_counts_rtn'] / df_final['row_counts_ord']

#sort to get the ans
df_final = df_final.sort_values(by = 'return_rate', ascending = False)
ans = df_final.iloc[0]

print(f"Size having the most return probability is: {ans['size']} with {ans['return_rate']:.2%}")

```
