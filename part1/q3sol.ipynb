```python
import pandas as pd

df_returns = pd.read_csv('../dataset/returns.csv')
df_products = pd.read_csv('../dataset/products.csv')
#merge the 2 dataframes
df = pd.merge(df_returns, df_products, on = 'product_id', how = 'left')
df_streetwear = df[df['category'] == 'Streetwear']
df_streetwear = df_streetwear.groupby('return_reason')['return_quantity'].sum().reset_index()

#sort to get the most common
df_streetwear = df_streetwear.sort_values(by = 'return_quantity', ascending= False)

#print ans
ans = df_streetwear.iloc[0]
print(f"Most common reason: {ans['return_reason']} - quantity: {ans['return_quantity']}")
```
