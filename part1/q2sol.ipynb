# Question 2
file: products.csv
cand: segments
target: (price - cogs)/price
calc: avg max

```python
import pandas as pd

df = pd.read_csv('../dataset/products.csv')
#set new col: gross_margin
df['gross_margin'] = (df['price'] - df['cogs'])/df['price']
#calc avg for each segment
segment_margin = df.groupby('segment')['gross_margin'].mean().reset_index()

#sort to get the max
segment_margin = segment_margin.sort_values(by = 'gross_margin', ascending= False)
print(segment_margin.iloc[0])
```
