```python
import pandas as pd

df_order = pd.read_csv('dataset/orders.csv')
df_order['order_date'] = pd.to_datetime(df_order['order_date'])
#filter to get the cand
df_del = df_order[df_order['order_status'] == 'delivered']
df_del = df_del.sort_values(['customer_id', 'order_date'])
inter_order_gap = df_del.groupby('customer_id')['order_date'].diff().dt.days.dropna()


```
