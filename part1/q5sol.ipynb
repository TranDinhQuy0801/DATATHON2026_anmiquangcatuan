# Question 5
file: order_items.csv
cand: promo_id != null
target: num of lines
calc: percentage

```python
import pandas as pd

df = pd.read_csv('../dataset/order_items.csv', dtype={'promo_id': str, 'promo_id_2': str})
#get cands
df_cand = df[(df['promo_id'].notna()) | (df['promo_id_2'].notna())]
#calc ans 
ans = (len(df_cand)/len(df))*100
#print ans
print(f"Percentage of ordered items had promotions is: {ans:.2f}%")
```
