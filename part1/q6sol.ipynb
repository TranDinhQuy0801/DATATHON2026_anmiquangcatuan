```python
import pandas as pd

# 1. Đọc dữ liệu từ thư mục dataset
customers = pd.read_csv('../dataset/customers.csv')
orders = pd.read_csv('../dataset/orders.csv')

# 2. Lọc khách hàng có age_group khác null
valid_customers = customers[customers['age_group'].notnull()]

# 3. Tính tổng số khách hàng theo từng nhóm tuổi
customer_counts = valid_customers.groupby('age_group')['customer_id'].nunique().reset_index()
customer_counts.columns = ['age_group', 'total_customers']

# 4. Kết nối với bảng orders để đếm số đơn hàng của những khách hàng này
customer_orders = pd.merge(valid_customers, orders, on='customer_id', how='inner')
order_counts = customer_orders.groupby('age_group')['order_id'].count().reset_index()
order_counts.columns = ['age_group', 'total_orders']

# 5. Kết hợp lại và tính toán tỷ lệ trung bình
final_stats = pd.merge(order_counts, customer_counts, on='age_group')
final_stats['avg_orders_per_customer'] = final_stats['total_orders'] / final_stats['total_customers']

# 6. Hiển thị kết quả sắp xếp giảm dần để tìm nhóm cao nhất
result = final_stats.sort_values(by='avg_orders_per_customer', ascending=False)

print("Thống kê số đơn hàng trung bình trên mỗi khách hàng theo nhóm tuổi:")
print(result[['age_group', 'avg_orders_per_customer']])
```
