```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

cust = pd.read_csv('../dataset/customers.csv')
orders = pd.read_csv('../dataset//orders.csv')

customer_orders = orders.groupby('customer_id').size().reset_index(name='order_count')
df = pd.merge(cust, customer_orders, on='customer_id')

plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='age_group', y='order_count', hue='gender', 
            estimator=sum, palette='viridis', 
            order=['18-24', '25-34', '35-44', '45-54', '55+'])

plt.title('Tổng số đơn hàng theo Nhóm tuổi và Giới tính')
plt.xlabel('Nhóm tuổi')
plt.ylabel('Tổng số đơn hàng đã mua')
plt.legend(title='Giới tính')
plt.show()
```
