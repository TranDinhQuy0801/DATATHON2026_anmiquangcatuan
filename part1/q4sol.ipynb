```python
import pandas as pd

df = pd.read_csv('../dataset/web_traffic.csv')
#group to get cand
df = df.groupby('traffic_source')['bounce_rate'].mean().reset_index()

#sort to get target ( min -> max)
df = df.sort_values(by = 'bounce_rate', ascending = True)
ans = df.iloc[0]
print("Traffic source with LEAST avg bounce rate: ", ans['traffic_source'])


```
