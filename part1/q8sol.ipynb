```python
import pandas as pd

df = pd.read_csv('../dataset/orders.csv')
#filter
df_cancelled = df[df['order_status'] == 'cancelled']
payment_count = df_cancelled['payment_method'].value_counts()

ans_method = payment_count.index[0]
ans_quantity = payment_count.iloc[0]
print("Payment method had the highest cancel quantity is:", ans_method, "with ", ans_quantity, "orders")
```
