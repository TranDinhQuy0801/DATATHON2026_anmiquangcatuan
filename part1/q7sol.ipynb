file: geography.csv sales.csv payments.csv
cand: region
target: revengue
calc: max
idea: payments + order (1:1) -> groupby('region') -> sum

```python
import pandas as pd

df_pay = pd.read_csv('../dataset/payments.csv')
df_ord = pd.read_csv('../dataset/orders.csv')
df_geo = pd.read_csv('../dataset/geography.csv')

#set new col from pay for ord -> get the revengue;
df_ord = pd.merge(df_ord, df_pay, on = 'order_id', how = 'left')

#groupby zip to get the sum of each zip
df_ord = df_ord.groupby('zip')['payment_value'].sum().reset_index()

#merge geo and ord -> get sales of region
df_geo = pd.merge(df_ord, df_geo, on = 'zip', how = 'left')

#groupby region -> get sales
df_geo = df_geo.groupby('region')['payment_value'].sum().reset_index()

#sort to get the highest
df_geo = df_geo.sort_values(by = 'payment_value', ascending = False)

ans = df_geo.iloc[0]
print("The region had the highest sales:", ans['region'], "with sale value:", ans['payment_value'])
```
