```python
import pandas as pd

df_pay = pd.read_csv('../dataset/payments.csv')

#group installments -> calc avg
df_installment = df_pay.groupby('installments')['payment_value'].mean().reset_index()

options = [1, 3, 6, 12]
df_options = df_installment[df_installment['installments'].isin(options)]

#sort to get max
df_options = df_options.sort_values(by='payment_value', ascending=False)

ans = df_options.iloc[0]
# In kết quả bằng tiếng Anh
print(f"The installment plan with the highest average payment value is {int(ans['installments'])} installments with avg value: {ans['payment_value']:.2f})")
```
